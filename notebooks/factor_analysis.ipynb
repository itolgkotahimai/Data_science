```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.decomposition import FactorAnalysis
from sklearn import datasets
```

### Factor analysis on iris dataset

```python
iris = datasets.load_iris()

print(iris.keys())
x = iris.data
y= iris.target
variables = iris.feature_names
factor = FactorAnalysis()
factor.fit(x)
df = pd.DataFrame(factor.components_, columns=variables)
print(df)
```
