```python
import numpy as np
import pandas as pd
import scipy
import matplotlib.pyplot as plt
from pylab import rcParams
import seaborn as sns
import urllib
import sklearn
from sklearn.neighbors import KNeighborsClassifier
from sklearn import neighbors, preprocessing
from sklearn.model_selection import train_test_split
from sklearn import metrics

np.set_printoptions(precision=4, suppress=True)

%matplotlib inline
rcParams['figure.figsize'] = 5,6

```

```python
#load data into dataframe
address = r'C:\Users\Lenovo\Downloads\Documents\Data science\Data_science\data\mtcars.csv'
data = pd.read_csv(address)
data.rename(columns={data.columns[0]: 'car_names'}, inplace= True)

#create more data frame
x_prime = data[['mpg', 'disp','hp','wt']].values
y = data['am'].values

#scaling our feature values
x = preprocessing.scale(x_prime)

#split data for train and test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state=17)

#train model
model = neighbors.KNeighborsClassifier()
model.fit(x_train, y_train)

#predict using trainded model
y_pred = model.predict(x_test)

#result
print(metrics.classification_report(y_test, y_pred))



```
